### Python Generator
Generator is a function which contains yield statement instead of return statement


```python
def gen_demo():
    yield "Hello"
    yield "World"
    yield 123
    yield np.array([1, 2, 3])
```

```python
iterator = gen_demo() # Here we are just creating an iterator object not calling it.
print(type(iterator))
print(iterator)
```

```python
# Here we are calling the iterator object to get the next value from the generator function -> 'next()'
for i in iterator:
    print(i)
```

**Property** -> the generator has already been exhausted — once you loop through a generator fully, it's used up and can't be iterated again. (Just re-execute the object creation of generator)

```python
def my_generator(n):
    for i in range(1, n + 1):
        yield i**2

gen = my_generator(10)
print(type(gen))
print(gen)
```

```python
# 1
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# 2
for i in gen:
    print(i)
```

```python
# 3 
print(next(gen))
print(next(gen))


for i in gen:
    print(i)
```

## Range function using Generator

```python
# Here the my_range is an iterator
def my_range(start, end):
    for i in range(start, end):
        yield i
```

```python
gen = my_range(1, 10)
for i in gen:
    print(i)

"""
This is an alternate to the above code.

for i in my_range(1, 10):
    print(i)
"""
```
