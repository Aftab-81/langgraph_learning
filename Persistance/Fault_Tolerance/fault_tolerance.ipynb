```python
from langgraph.graph import StateGraph
from typing import TypedDict, Self
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from langgraph.checkpoint.memory import InMemorySaver
import time
load_dotenv()
```

```python
# Define the state graph
class CrashState(TypedDict):
    input: str
    step1: str
    step2: str
    step3: str
```

```python
# Task 1
def step_1(state: CrashState) -> Self:
    print("✅ Step 1 executed..")
    return {"step1": "done", "input": state["input"]}

# Task 2
def step_2(state: CrashState) -> Self:
    print("⏳ Step 2 hanging... now manually interrupt from the notebook toolbar (STOP button)")
    time.sleep(30) # Simulate a long-running task that may hang or crash
    return {"step2": "done"}

# Task 3
def step_3(state: CrashState) -> Self:
    print("✅ Step 3 executed..")
    return {"step3": "done"}
```

```python
# Build graph
graph = StateGraph(CrashState)

# Add nodes
graph.add_node("step1", step_1)
graph.add_node("step2", step_2)
graph.add_node("step3", step_3)

# Add edges
graph.set_entry_point("step1")
graph.add_edge("step1", "step2")
graph.add_edge("step2", "step3")
graph.set_finish_point("step3")

# Create a checkpoint saver
memory = InMemorySaver()

# Compilation
workflow = graph.compile(checkpointer = memory)
workflow
```

```python
initial_input = {"input": "art"}
config = {"configurable": {"thread_id": "1"}}
try:
    print("▶️ Running graph manually interrupt the execution in the middle of step2 to test fault tolerance")
    workflow.invoke(initial_input, config = config)
except KeyboardInterrupt: 
    print("❌ Execution interrupted. Saving state...")
```

```python
workflow.get_state(config = config)
```

```python
list(workflow.get_state_history(config = config))
```

```python
# Here instead of passing 'initial_state' we pass 'None' means I want to resume the workflow where it last left off. The workflow will automatically pick up from the last saved state.
final_state = workflow.invoke(None, config = config)

# Here the workflow will start from step2
print(final_state)
```
